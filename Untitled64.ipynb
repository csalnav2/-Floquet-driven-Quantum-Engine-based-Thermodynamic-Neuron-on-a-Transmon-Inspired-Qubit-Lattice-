```python
 jfrom getpass import getpass
from pathlib import Path
import json
import os


openai_key = getpass(
    "OpenAI API key: "
).strip()

ngrok_token = getpass(
    "NEW ngrok authtoken—not the ngrok API key: "
).strip()


if not openai_key:
    raise ValueError("OPENAI_API_KEY cannot be empty.")

if not ngrok_token:
    raise ValueError("NGROK_AUTHTOKEN cannot be empty.")

if any(character.isspace() for character in ngrok_token):
    raise ValueError(
        "The ngrok authtoken contains whitespace. "
        "Paste only the token itself."
    )


# Clear stale values from older versions of the notebook.
for old_name in (
    "DASHBOARD_USERNAME",
    "DASHBOARD_PASSWORD",
    "OPENAI_ORG_ID",
    "OPENAI_PROJECT_ID",
    "DASHBOARD_ACCESS_TOKEN",
):
    os.environ.pop(old_name, None)


env_path = Path("/content/.env")

env_path.write_text(
    (
        f"OPENAI_API_KEY={json.dumps(openai_key)}\n"
        f"NGROK_AUTHTOKEN={json.dumps(ngrok_token)}\n"
    ),
    encoding="utf-8",
)

try:
    os.chmod(env_path, 0o600)
except OSError:
    pass


print("Saved:", env_path)
print("OPENAI_API_KEY: SET")
print("NGROK_AUTHTOKEN: SET")
print("No credential values were displayed.")
```
